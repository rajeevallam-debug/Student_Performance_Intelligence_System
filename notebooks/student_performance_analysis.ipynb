# Student Performance Intelligence System

This notebook analyzes sample student marks and summarizes performance across subjects.


```python
import numpy as np
import pandas as pd

df = pd.read_csv('../data/student_marks.csv')
df['Total'] = df[['Maths', 'Science', 'English', 'History']].sum(axis=1)
df['Average'] = df[['Maths', 'Science', 'English', 'History']].mean(axis=1)
print(df)
print('\nSubject averages:')
print(df[['Maths', 'Science', 'English', 'History']].mean())
print('\nTop performers:')
print(df.sort_values('Total', ascending=False)[['Student', 'Total']])
```
