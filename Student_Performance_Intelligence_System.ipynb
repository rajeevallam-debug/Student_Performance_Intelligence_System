```python
import numpy as np
```

Create our student **DATASET:**





```python
students = np.array([
    [85, 78, 92],
    [65, 72, 68],
    [95, 91, 89],
    [45, 52, 48],
    [76, 81, 79]
])

print(students)
```

**Think of this as a table:**
Students with subject names called Python,Statistics,DataVisualization and thier Marks


**UNDERSTAND THE ARRAY:**

```python
print(students.shape)
```

**THIS MEANS:**
5 rows
3 columns

```python
print(students.ndim)
```

That's because our data is a **2-dimensional array**.

```python
print(students.size)
```

5 × 3 = 15;5-Rows,3-Coloumns


**Access individual data:**

Let's access Student 1's Python mark:

```python
print(students[0, 0])
```

EXAMPLES:

```python
print(students[2, 1])


print(students[4, 2])
```

**Calculate total marks:**

Now we start doing actual analysis.


We want each student's total.

```python
total_marks = np.sum(students, axis=1)

print(total_marks)
```

**Important concept:** axis=1

Student 1 → 85 + 78 + 92

Student 2 → 65 + 72 + 68

...

**Calculate average marks:**

```python
average_marks = np.mean(students, axis=1)

print(average_marks)
```

```python
average_marks = np.round(np.mean(students, axis=1), 2)

print(average_marks)
```

**Find the highest-performing student:**

```python
highest_index = np.argmax(average_marks)

print(highest_index)
```

Let's make the output more understandable:

```python
print("Highest performing student: Student", highest_index + 1)
print("Average:", average_marks[highest_index])
```

**Find the lowest-performing student:**

```python
lowest_index = np.argmin(average_marks)

print("Lowest performing student: Student", lowest_index + 1)
print("Average:", average_marks[lowest_index])
```

This uses:

np.argmax() → largest value

np.argmin() → smallest value

**Find students who failed:**

```python
failed = students < 50

print(failed)
```

True means condition satisfies.

False means condition not satisfies.

Now we can find students who have **at least one failed subject:**

```python
failed_students = np.any(students < 50, axis=1)

print(failed_students)
```

**Subject-wise analysisS:**

Python Marks:

```python
python_marks = students[:, 0]

print("Python Marks:", python_marks)
```

Statistics marks:

```python
statistics_marks = students[:, 1]

print("Statistics Marks:", statistics_marks)
```

Data Visualization marks:

```python
data_visualization_marks = students[:, 2]

print("Data Visualization Marks:", data_visualization_marks)
```

**Calculate subject averages:**

```python
python_avg = np.mean(python_marks)
statistics_avg = np.mean(statistics_marks)
data_visualization_avg = np.mean(data_visualization_marks)

print("Python Average:", python_avg)
print("Statistics Average:", statistics_avg)
print("Data Visualization Average:", data_visualization_avg)
```

Instead of calculating each column separately, NumPy can calculate all subject averages at once:

```python
subject_averages = np.mean(students, axis=0)

print(subject_averages)
```

**Find the highest mark in each subject:**

```python
highest_subject_marks = np.max(students, axis=0)

print(highest_subject_marks)
```

This tells us:

Python              → 95

Statistics          → 91

Data Visualization  → 92

**Calculate standard deviation:**

```python
std_marks = np.std(students, axis=0)

print(std_marks)
```

**Simple interpretation**


If standard deviation is:

Small

↓

Marks are relatively close together

Large

↓





Marks are more spread out

**Create student rankings 🏆:**

```python
ranking = np.argsort(average_marks)[::-1]

print("STUDENT RANKING")
print("----------------")

for rank, index in enumerate(ranking, start=1):
    print(
        "Rank", rank,
        "- Student", index + 1,
        "- Average:", average_marks[index]
    )
```
